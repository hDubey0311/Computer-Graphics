```python
import matplotlib.pyplot as plt
x = 5
y = 8
#Plotting the point
plt.scatter(x,y,color='red', s = 200)
#Display the co-ordinate near the point
plt.text(x+0.2,y+0.2,f'({x},{y})')
#Set graph limits
plt.xlim(0,10)
plt.ylim(0,10)
#Display grid
plt.grid(True)
#Label the axes
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
#Dispaly graph title
plt.title("Plotting the single point")
#Display student's name
plt.text(5,9.5,"Hitesh Dubey",fontsize=12,color="blue", ha="center")
#Display the graph
plt.show()
```

```python
import matplotlib.pyplot as plt
a = int(input("Enter the a-coordinate: "))
b = int (input("Enter the b-coordinate: "))
#Plotting the point
plt.scatter(a,b,color='red', s = 200)
#Display the co-ordinate near the point
plt.text(a+0.2,b+0.2,f'({a},{b})')
#Set graph limits
plt.xlim(0,10)
plt.ylim(0,10)
#Display grid
plt.grid(True)
#Label the axes
plt.xlabel("X-axis")#Axis are the constant according to the lib (plt.alabel("A-axis")) = error, cause = (.alabel), solution = (.xlabel), Reason = the permanent label given to axis is 'x' in the library we are just giving the name to 'x axis' as 'a axis'
plt.ylabel("Y-axis")#Axis are the constant according to the lib (plt.blabel("B-axis")) = error, cause = (.blabel), solution = (.ylabel), Reason = the permanent label given to axis is 'y' in the library we are just giving the name to 'y axis' as 'b axis'
#Dispaly graph title
plt.title("Plotting a point using user defined coordinates")
#Display student's name
plt.text(5,9.5,"Hitesh Dubey",fontsize=12,color="blue", ha="center")
#Display the graph
plt.show()
```

```python
import matplotlib.pyplot as plt
# Coordinates of the first point
x1 = 3
y1 = 4
# Coordinates of the second point
x2 = 8
y2 = 7
# Plot the first point
plt.scatter(x1, y1, color='red', s=100)
# Plot the second point
plt.scatter(x2, y2, color='blue', s=100)
# Display the coordinates
plt.text(x1 + 0.2, y1 + 0.2, f'({x1}, {y1})')
plt.text(x2 + 0.2, y2 + 0.2, f'({x2}, {y2})')
#Set graph limits
plt.xlim(0, 10)
plt.ylim(0, 10)
#Display grid
plt.grid(True)
#Label the axes
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
# Display graph title
plt.title("Plotting Two Points")
# Display student's name
plt.text(5, 9.5, "Hitesh Dubey", fontsize=12, color="blue", ha="center")
#Show the graph
plt.show()
```

```python
import matplotlib.pyplot as plt
# Coordinates of the first, second and third point
x1 = 8
y1 = 9
x2 = 3
y2 = 4
x3 = 6
y3 = 1
# Plot the first point
plt.scatter(x1, y1, color='red', s=100)
# Plot the second point
plt.scatter(x2, y2, color='blue', s=100)
# Plot the third point
plt.scatter(x3, y3, color='green', s=100)
# Display the coordinates
plt.text(x1 + 0.2, y1 + 0.2, f'({x1}, {y1})')
plt.text(x2 + 0.2, y2 + 0.2, f'({x2}, {y2})')
plt.text(x3 + 0.2, y3 + 0.2, f'({x3}, {y3})')
# Set graph limits
plt.xlim(0, 10)
plt.ylim(0, 10)
#Display grid
plt.grid(True)
#Label the axes
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
# Display graph title
plt.title("Plotting Three Points Using Different Colors")
# Display student's name
plt.text(5, 9.5, "Hitesh Dubey", fontsize=12, color="blue", ha="center")
# Show the graph
plt.show()
```

```python
import matplotlib.pyplot as plt
# Accept coordinates from the user
x = int(input("Enter x-coordinate: "))
y = int(input("Enter y-coordinate: "))
# Accept color from the user
color = input("Enter point color: ").lower()
# Plot the point
plt.scatter(x, y, color=color, s=100)
# Display the coordinate near the point
plt.text(x + 0.2, y + 0.2, f'({x}, {y})')
# Set graph limits
plt.xlim(0, 10)
plt.ylim(0, 10)
# Display grid
plt.grid(True)
#Label the axes
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
# Display graph title
plt.title("Plotting a Point Using User-Defined Coordinates and Color")
# Display student's name
plt.text(5, 9.5, "Hitesh Dubey", fontsize=12, color="blue", ha="center")
# Show the graph
plt.show()
```
