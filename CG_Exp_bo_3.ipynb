```python
import matplotlib.pyplot as plt
#Function to implement Bresenham's Line Algorithm (Ax > Ay)
def bresenham_line (x1, y1, x2, y2):
  dx = x2 - x1
  dy = y2 - y1
  # Validate input
  if dx <= 0 or dy < 0 or dx <= dy:
    print("\nError: Invalid input!")
    print("This program works only when:")
    print("1. X2 > X1 (Left-to-Right)")
    print("2. y2 ≥ Y1 (Positive Slope)")
    print("3. Ax > Ay (Slope < 1)")
    return None, None
  # Initial Decision Parameter
  p = 2 * dy - dx
  x = x1
  y = y1
  x_points = [x]
  y_points = [y]
  print("\nStep\tCurrent Pixel\tP\tDecision\tNext Pixel")
  step = 0
  while x < x2:
    if p < 0:
      decision = "E"
      p = p+ (2 * dy)
      x = x + 1
    else:
      decision = "NE"
      p= p+ (2 * dy) - (2 * dx)
      x = x+ 1
      y = y + 1
    print (f"{step}\t({x_points[-1]}, {y_points[-1]})\t{p}\t{decision}\t({x},{y})")
    x_points.append (x)
    y_points.append (y)
    step += 1
  return x_points, y_points
# Main Program
print("========== Bresenham's Line Drawing Algorithm=========")
print("Condition: Ax Ay (Slope < 1)\n")
x1 = int (input("Enter x1: "))
y1 = int (input("Enter yl: "))
x2 = int (input("Enter x2: "))
y2 = int (input("Enter y2: "))
x_points, y_points = bresenham_line (x1, y1, x2, y2)
# Plot only if valid input
if x_points is not None:
  student_name = input("\nEnter your Name: ")
  plt.figure (figsize=(6, 6))
# Plot pixels
plt.scatter (x_points, y_points, color="blue", s=100)
# Join pixels for better visualization
plt.plot (x_points, y_points, linestyle="--")
#Display pixel coordinates
for x, y in zip(x_points, y_points):
  plt.text (x + 0.1, y + 0.1, f"({x},{y})", fontsize=9)
plt.title(f"Bresenham's Line Drawing Algorithm\nStudent:{student_name}")
plt.xlabel("X-axis")
plt.ylabel ("Y-axis")
plt.grid (True)
plt.axis ("equal")
plt.show ()
```
