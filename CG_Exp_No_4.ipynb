```python
import matplotlib.pyplot as plt

# Function to generate 8-way symmetric points
def plot_circle_points(xc, yc, x, y, points):
    symmetric_points = [
        (xc + x, yc + y),
        (xc - x, yc + y),
        (xc + x, yc - y),
        (xc - x, yc - y),
        (xc + y, yc + x),
        (xc - y, yc + x),
        (xc + y, yc - x),
        (xc - y, yc - x)
    ]
    points.extend(symmetric_points)

# Midpoint Circle Algorithm
def midpoint_circle(xc, yc, r):
    x = 0
    y = r
    p = 1 - r
    points = []

    while x <= y:
        plot_circle_points(xc, yc, x, y, points)
        if p < 0:
            p = p + 2 * x + 3
        else:
            p = p + 2 * (x - y) + 5
            y -= 1
        x += 1
    return points

# Main Program
student_name = input("Enter Your Name: ")
all_points = []
print("\nEnter Details of Two Circles")

for i in range(2):
    print(f"\nCircle {i+1}")
    xc = int(input("Enter Center X: "))
    yc = int(input("Enter Center Y: "))
    r = int(input("Enter Radius: "))
    circle = midpoint_circle(xc, yc, r)
    all_points.extend(circle)

# Remove duplicate points
all_points = list(set(all_points))

# Separate X and Y coordinates
x_values = []
y_values = []
for point in all_points:
    x_values.append(point[0])
    y_values.append(point[1])

# Plot the circles
plt.figure(figsize=(8, 8))
plt.scatter(x_values, y_values, color='blue', s=40)
plt.title(f"Two Circles using Midpoint Circle Algorithm\nStudent: {student_name}")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.axis('equal')
plt.show()
```
