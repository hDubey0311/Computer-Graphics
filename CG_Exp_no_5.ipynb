```python
import matplotlib.pyplot as plt

def midpoint_ellipse(xc, yc, rx, ry):
    x = 0
    y = ry
    points = []

    # Initial values
    dx = 2 * ry * ry * x
    dy = 2 * rx * rx * y

    # Region 1 decision parameter
    p1 = (ry * ry) - (rx * rx * ry) + (0.25 * rx * rx)

    # Function for 4-way symmetry
    def add_points(x, y):
        points.append((xc + x, yc + y))
        points.append((xc - x, yc + y))
        points.append((xc + x, yc - y))
        points.append((xc - x, yc - y))

    # Region 1
    while dx < dy:
        add_points(x, y)
        if p1 < 0:
            x = x + 1
            dx = 2 * ry * ry * x
            p1 = p1 + dx + (ry * ry)
        else:
            x = x + 1
            y = y - 1
            dx = 2 * ry * ry * x
            dy = 2 * rx * rx * y
            p1 = p1 + dx - dy + (ry * ry)

    # Region 2 decision parameter
    p2 = ((ry * ry) * ((x + 0.5) ** 2)) + ((rx * rx) * ((y - 1) ** 2)) - ((rx * rx) * (ry * ry))

    # Region 2
    while y >= 0:
        add_points(x, y)
        if p2 > 0:
            y = y - 1
            dy = 2 * rx * rx * y
            p2 = p2 + (rx * rx) - dy
        else:
            x = x + 1
            y = y - 1
            dx = 2 * ry * ry * x
            dy = 2 * rx * rx * y
            p2 = p2 + dx - dy + (rx * rx)

    return points

# Input
name = input("Enter student's name: ")
rx = int(input("Enter x-radius (rx): "))
ry = int(input("Enter y-radius (ry): "))

# Center of ellipse
xc = 0
yc = 0

# Generate ellipse points
points = midpoint_ellipse(xc, yc, rx, ry)

# Separate x and y coordinates
x_values = [point[0] for point in points]
y_values = [point[1] for point in points]

# Plot
plt.figure(figsize=(8, 6))
plt.scatter(x_values, y_values, s=20)

# Coordinate axes
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)

plt.grid(True)
plt.axis("equal")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title(f"Midpoint Ellipse Algorithm\nStudent: {name} | rx={rx} ry={ry}")
plt.show()
```
