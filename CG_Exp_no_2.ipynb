```python
import matplotlib.pyplot as plt
#Input the starting point
x1 = int(input("Enter x1 value:"))
y1 = int(input("Enter y1 value:"))
#Input the ending point
x2 = int(input("Enter x2 value:"))
y2 = int(input("Enter y2 value:"))
#Displaying x1,y1 & x2,y2
print(f"(x1,y1) = {(x1,y1)}")
print(f"(x2,y2) = {(x2,y2)}")
#Calculate the difference
dx = x2 - x1
dy = y2 - y1
#Find number of steps
steps = max(abs(dx),abs(dy))
#Calculate the x & y increment
x_increment = dx / steps
y_increment = dy / steps
# Initialize the starting point
x = x1
y = y1
# Lists to store the generated points
x_points = []
y_points = []
print("\nGenerated...")
print("-------------...")
print("Step\tX\tY")
i = 0
# Generate the intermediate points
while i <= steps:
  print(f"{i}\t{round(x)}\t{round(y)}")
  x_points.append(round(x))
  y_points.append(round(y))
  x = x + x_increment
  y = y + y_increment
  i += 1
# Plot the generated points
plt.scatter(x_points, y_points, color="red", s=80)
# Join the points
plt.plot(x_points, y_points)
# Display the coordinates
for i in range(len(x_points)):
    plt.text(x_points[i] + 0.1,
             y_points[i] + 0.1,
             f"({x_points[i]}, {y_points[i]})",
             fontsize=8)
# Set graph limits
plt.xlim(min(x_points) - 1, max(x_points) + 1)
plt.ylim(min(y_points) - 1, max(y_points) + 1)
# Display grid
plt.grid(True)
# Label the axes
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
# Display graph title
plt.title("DDA Line Drawing Algorithm : Hitesh Dubey")
# Display the graph
plt.show()
```

```python
import matplotlib.pyplot as plt

import matplotlib.pyplot as plt

# ------------------------- Line 1 -------------------------

# Input the starting point of line 1
x1 = int(input("Enter x1: "))
y1 = int(input("Enter y1: "))

# Input the ending point of line 1
x2 = int(input("Enter x2: "))
y2 = int(input("Enter y2: "))

# ------------------------- Line 2 -------------------------

# Input the starting point of line 2
a1 = int(input("Enter a1: "))
b1 = int(input("Enter b1: "))

# Input the ending point of line 2
a2 = int(input("Enter a2: "))
b2 = int(input("Enter b2: "))

# ------------------------- DDA for Line 1 -------------------------

# Calculate the differences
dx = x2 - x1
dy = y2 - y1

# Determine the number of steps
steps1 = max(abs(dx), abs(dy))

# Calculate the increments
x_increment = dx / steps1
y_increment = dy / steps1

# Initialize the starting point
x = x1
y = y1

# Lists to store the generated points
x_points = []
y_points = []

i = 0

# Generate the intermediate points
while i <= steps1:

    x_points.append(round(x))
    y_points.append(round(y))

    x = x + x_increment
    y = y + y_increment

    i += 1

# ------------------------- DDA for Line 2 -------------------------

# Calculate the differences
da = a2 - a1
db = b2 - b1

# Determine the number of steps
steps2 = max(abs(da), abs(db))

# Calculate the increments
a_increment = da / steps2
b_increment = db / steps2

# Initialize the starting point
a = a1
b = b1

# Lists to store the generated points
a_points = []
b_points = []

j = 0

# Generate the intermediate points
while j <= steps2:

    a_points.append(round(a))
    b_points.append(round(b))

    a = a + a_increment
    b = b + b_increment

    j += 1

# ------------------------- Plot Line 1 -------------------------

plt.scatter(x_points, y_points, color="red", s=80)
plt.plot(x_points, y_points, color="red")

# ------------------------- Plot Line 2 -------------------------

plt.scatter(a_points, b_points, color="green", s=80)
plt.plot(a_points, b_points, color="green")

# ------------------------- Set Graph Limits -------------------------

all_x = x_points + a_points
all_y = y_points + b_points

plt.xlim(min(all_x) - 1, max(all_x) + 1)
plt.ylim(min(all_y) - 1, max(all_y) + 1)

# ------------------------- Display Grid -------------------------

plt.grid(True)

# Label the axes
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

# Display graph title
plt.title("DDA Line Drawing Algorithm : Hitesh Dubey")

# Display the graph
plt.show()
```
